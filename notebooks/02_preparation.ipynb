```python
import pandas as pd
df = pd.read_csv(r"D:\project\HR_Attrition_Analysis\data\HR-Employee-Attrition.csv")
```

```python
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18", "EmployeeNumber"])
df.shape    # should be (1470, 31)
```

```python
df["AttritionFlag"] = df["Attrition"].map({"Yes": 1, "No": 0})
df["AttritionFlag"].mean()    # should be about 0.161
```

```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[17, 25, 35, 45, 55, 100],
                        labels=["18-25", "26-35", "36-45", "46-55", "56+"])

df["IncomeLevel"] = pd.cut(df["MonthlyIncome"], bins=[0, 3000, 6000, 10000, 20000],
                           labels=["Low", "Medium", "High", "Very High"])

df["TenureBand"] = pd.cut(df["YearsAtCompany"], bins=[-1, 2, 5, 10, 50],
                          labels=["0-2 yrs", "3-5 yrs", "6-10 yrs", "11+ yrs"])
```
